Write a program to print a sequence of numbers from 1 to 9 using a while loop. The program should initialize a starting number, use a while loop to iterate through the numbers, print each number in the sequence, and then confirm that the code has executed successfully by printing a confirmation message.


INPUT
No direct input is required.


CONSTRAINTS
The program must initialize a variable x with the integer value 1.

Use a while loop to iterate as long as x is less than 10.

In each iteration, print the current value of x.

Increment x by 1 after each iteration.

After the loop completes, the program must print the message "Code executed successfully !" directly.

Include comments in the code to explain its functionality.


OUTPUT
The console should display the numbers from 1 to 9, each on a new line, followed by the confirmation message. For example:


1

2

3

4

5

6

7

8

9

Code executed successfully !


```python
x = 1

while x <= 9:
    print(x)
    x += 1
print("Code executed successfully !")   
```

Write a program to iterate through a predefined range of numbers using a for loop. The program should print each number in the sequence until it encounters a specific number, at which point it should terminate the loop using a break statement. After exiting the loop, the program should confirm that the code has executed successfully by printing a confirmation message.


INPUT
No direct input is required.


CONSTRAINTS
The program must use a for loop to iterate through numbers in the range from 1 to 4 (inclusive).

Within the loop:

If the current number is equal to 3, the program should terminate the loop using the break statement.

Otherwise, it should print the current number.

After the loop concludes (whether normally or via break), the program must print the message "Code executed successfully !" directly.

Include comments in the code to explain its functionality.


OUTPUT
The console should display the numbers printed before the loop is terminated, followed by the confirmation message. For example:


1

2

Code executed successfully !

```python
for i in range(1,4):
    if i == 3:
        break
```
