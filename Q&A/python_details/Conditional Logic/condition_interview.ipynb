```python
def parse_bool(value):
    if value is None:
        return None

    text = str(value).strip().lower()

    if text in ("true", "yes", "1"):
        return True
    elif text in ("false", "no", "0"):
        return False
    elif text in ("", "na", "n/a", "null"):
        return None
    else:
        raise ValueError(f"Unexpected Boolean value: {value!r}")
```
